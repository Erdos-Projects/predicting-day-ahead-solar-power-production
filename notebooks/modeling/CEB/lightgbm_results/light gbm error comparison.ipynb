Light GBM error comparison

```python
import pandas as pd

df10 = pd.read_csv('10_None.csv')
df50 = pd.read_csv('50_None.csv')
df51 = pd.read_csv('51_None.csv')
```

System 10

```python
df = df10

summary = pd.DataFrame({
    "mean": df.mean(axis=0),
    "median": df.median(axis=0),
    "max": df.max(axis=0),
    "std": df.std(axis=0)
})

summary = summary.reset_index()
summary = summary.rename(columns={"index": "hyperparameters"})

# sort by mean
summary = summary.sort_values("mean")

summary.head(25)

#48
#but 56 isn't too bad
```
